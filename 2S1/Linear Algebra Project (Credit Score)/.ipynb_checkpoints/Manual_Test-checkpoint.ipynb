## Convert Data Type

### Float Type

```python
indices_to_float = [0, 2, 3, 4, 5, 6, 7, 12, 14, 15, 18, 19, 9, 10, 11, 16, 21]
```

```python
converted_data = []
for i in range(len(split_list)):
    if i in indices_to_float:
        converted_data.append(float(split_list[i]))  
    else:
        converted_data.append(split_list[i])  
```

```python
print(converted_data)
```

### Strign Type

#### Occupation

```python
Occupation_Encoder = joblib.load('encoder/Occupation_Encoder.pkl')
converted_data[1] = Occupation_Encoder.transform([converted_data[1]])[0]
```

```python
print(converted_data)
```

#### Type_of_Loan

```python
Type_of_Loan_Encoder = joblib.load('encoder/Type_of_Loan_Encoder.pkl')
converted_data[8] = Type_of_Loan_Encoder.transform([converted_data[8]])[0]
```

```python
print(converted_data)
```

#### Credit_Mix

```python
Credit_Mix_Encoder = joblib.load('encoder/Credit_Mix_Encoder.pkl')
converted_data[13] = Credit_Mix_Encoder.transform([[converted_data[13]]])[0][0]
```

```python
print(converted_data)
```

#### Payment_of_Min_Amount

```python
Payment_of_Min_Amount_Encoder = joblib.load('encoder/Payment_of_Min_Amount_Encoder.pkl')
converted_data[17] = Payment_of_Min_Amount_Encoder.transform([converted_data[17]])[0]
```

```python
print(converted_data)
```

#### Payment_Behaviour

```python
Payment_Behaviour_Encoder = joblib.load('encoder/Payment_Behaviour_Encoder.pkl')
payment_behaviour = [[converted_data[20]]]  
encoded_value = Payment_Behaviour_Encoder.transform(payment_behaviour)
converted_data[20] = encoded_value[0][0]
```

```python
print(converted_data)
```

## Model Prediction

```python
model = keras.models.load_model('model/Credit_Score_Classification_Model.keras')
# model = keras.models.load_model('model/save.keras')
```

```python
input_data = pd.DataFrame([converted_data], columns=[
    'Age', 'Occupation', 'Annual_Income', 'Monthly_Inhand_Salary', 'Num_Bank_Accounts', 
    'Num_Credit_Card', 'Interest_Rate', 'Num_of_Loan', 'Type_of_Loan', 'Delay_from_due_date', 
    'Num_of_Delayed_Payment', 'Changed_Credit_Limit', 'Num_Credit_Inquiries', 'Credit_Mix', 'Outstanding_Debt', 
    'Credit_Utilization_Ratio', 'Credit_History_Age', 'Payment_of_Min_Amount', 'Total_EMI_per_month', 'Amount_invested_monthly', 
    'Payment_Behaviour', 'Monthly_Balance'
])
```

```python
predictions = model.predict(input_data)
```

```python
predicted_classes = np.argmax(predictions, axis=1)
```

```python
mapping = {
    0: 'Good',
    1: 'Standard',
    2: 'Poor'
}
```

```python
string_predict = [mapping[pred] for pred in predicted_classes]
```

```python
print("Predicted Labels : ", string_predict[0])
print("Actual Labels    : ", result)
predicted_label = string_predict[0]  
if predicted_label.lower() == result.lower():  
    print("Correct!")
else:
    print("Incorrect!")  
```
